### Models
![](images/best_fit_line.png){width=60%}

Regression can be used to model this relationship using a best-fitting line. A model is a mathematical equation that represents a causal relationship in the data. That is, we should understand the value of the independent variable to cause the value of the dependent variable, but this deterministic relationship is partially hidden by additive noise. This is represented by the equation $y = f(x) + \epsilon$, where $y$ is the observed value, $\epsilon$ is additive noise, and $f(x)$ is a deterministic equation. Therefore, the model predicts $y$ using $\hat{y} = f(x)$, where $\hat{y}$ is the predicted value of $y$.

### Ordinary Least Squares
This is a type of regression algorithm with the goal being to minimize the Residual Sum of Squares (RSS). A residual ($e$) is calculated by subtracting the predicted value from the actual value: $e_i = y_i - \hat{y}_i$

The residuals go in the vertical direction only, which implies that the x values caused the y values. When calculating the RSS, the residuals are squared, making outliers have a large influence on the sum. The formula for RSS is as follows: 
$$RSS = \sum_{i=1}^{n} e_i^2 = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

### Correlation
Correlation measures the strength of a relationship,  and stronger relationships have a higher correlation. Pearson's correlation coefficient ($r$) indicates how much the relationship between x and y is deterministic and how much is random noise:
$$r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}$$
In this equation, $x_i$ and  $y_i$ are individual data points, $\bar{x}$ is the mean of all $x$ values, and $\bar{y}$ is the mean of all $y$ values. Pearson's $r$ values always fall between -1 and 1. The sign indicates the direction, and the absolute number indicates the strength. That is, $r$ values closer to -1 or 1 mean a strong relationship, and an $r$ value closer to 0 represents a weak relationship.

![](images/correlations.png)

When Pearson's $r$ is squared it becomes the Coefficient of Determination: 

$$R^2 = (r)^2$$
$R^2$ can be interpreted as the percent variance explained by the model; in other words, it tells you how much of the variation in $Y$ can be predicted from knowing $X$. $R^2$ ranges from 0 to 1, and the closer it is to 1 the more of the variance the model explains.
